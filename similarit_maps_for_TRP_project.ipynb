```python
from rdkit import Chem
from rdkit.Chem import DataStructs, Draw , rdFingerprintGenerator
ref1=Chem.MolFromSmiles('COC1=CC(=C(C=C1)OC)NC(=O)C2=C(N(N=N2)CC3=C(C=C(C=C3)Br)F)N') #GNE-551 reference for TRPA1
ref2=Chem.MolFromSmiles(r'CO/N=C(/C1=CSC(=N1)N)\C(=O)N[C@@H](CO)[C@@H]2N=C(C(=C)CS2)C(=O)O') #TC-I reference for TRPM8
mol1=Chem.MolFromSmiles('CC1(CC(=O)C2=C(C=C(C=C2O1)O[C@H]3[C@@H]([C@H]([C@@H]([C@H](O3)COC(=O)C4=CC[C@@H](CC4)C(C)(C)O)O)O)O)O)O') # cypellocarpin
mol2=Chem.MolFromSmiles('C1=CC(=C(C=C1C[C@H](C(=O)O)OC(=O)/C=C/C2=CC(=C(C=C2)O)O)O)O') # rosmarinic
mol3=Chem.MolFromSmiles('OC1=C(C(CC(C)C)=O)C(O)=C(CC2=C(O)C(C(CC(C)C)=O)=C(O)C(C=O)=C2O)C(O)=C1C') # jensenal
mol4=Chem.MolFromSmiles('OC1=C2C(OC(C4=CC(O)=C(O)C=C4)=C(O[C@H]3[C@H](O)[C@@H](O)[C@H](CO)O3)C2=O)=CC(O)=C1') # Avicularin

gen=rdFingerprintGenerator.GetMorganGenerator(radius=2,fpSize=2048)
ref1_fp=gen.GetFingerprint(ref1)
ref2_fp=gen.GetFingerprint(ref2)
mol1_fp=gen.GetFingerprint(mol1)
mol2_fp=gen.GetFingerprint(mol2)
mol3_fp=gen.GetFingerprint(mol3)
mol4_fp=gen.GetFingerprint(mol4)


sim1=DataStructs.FingerprintSimilarity(ref1_fp,mol1_fp)
print(f"similarity between reference GNE-551 and cypellocarpin is : {sim1}")

from rdkit.Chem import Draw
from rdkit.Chem.Draw import IPythonConsole
from rdkit.Chem.Draw import SimilarityMaps
drawer=Draw.MolDraw2DSVG(600 ,600)
fig , maxweight = SimilarityMaps.GetSimilarityMapForFingerprint( ref1 , mol1 , SimilarityMaps.GetMorganFingerprint , draw2d=drawer)
drawer.FinishDrawing()
svg=drawer.GetDrawingText()
from IPython.display import SVG
display(SVG(svg))
```

```python
sim2=DataStructs.FingerprintSimilarity(ref1_fp,mol2_fp)
print(f"similarity between reference GNE-551 and rosmarinic is : {sim2}")

drawer=Draw.MolDraw2DSVG(600 ,600)
fig , maxweight = SimilarityMaps.GetSimilarityMapForFingerprint( ref1 , mol2 , SimilarityMaps.GetMorganFingerprint , draw2d=drawer)
drawer.FinishDrawing()
svg=drawer.GetDrawingText()
from IPython.display import SVG
display(SVG(svg))

```

```python
sim3=DataStructs.FingerprintSimilarity(ref2_fp,mol3_fp)
print(f"similarity between reference TC-I and jensenal  is : {sim3}")

drawer=Draw.MolDraw2DSVG(600 ,600)
fig , maxweight = SimilarityMaps.GetSimilarityMapForFingerprint( ref2 , mol3 , SimilarityMaps.GetMorganFingerprint , draw2d=drawer)
drawer.FinishDrawing()
svg=drawer.GetDrawingText()
from IPython.display import SVG
display(SVG(svg))
```

```python
sim4=DataStructs.FingerprintSimilarity(ref2_fp,mol4_fp)
print(f"similarity between reference TC-I and Avicularin  is : {sim4}")

drawer=Draw.MolDraw2DSVG(600 ,600)
fig , maxweight = SimilarityMaps.GetSimilarityMapForFingerprint( ref1 , mol4 , SimilarityMaps.GetMorganFingerprint , draw2d=drawer)
drawer.FinishDrawing()
svg=drawer.GetDrawingText()
from IPython.display import SVG
display(SVG(svg))
```
